```python
import pandas as pd
from sklearn.metrics import accuracy_score
```

```python
df_gold = pd.read_csv("../data/gold/salary_gold.csv")
df_pred = pd.read_csv("../data/gold/salary_pred.csv")
```

```python
df_pred["min_salary"] = df_pred["min_salary"].astype(int).astype(str)
df_gold["min_salary"] = df_gold["min_salary"].astype(int).astype(str)
```

```python
accuracy_score(y_true=df_gold["min_salary"], y_pred=df_pred["min_salary"])
```

```python
df_jobs = pd.read_csv("data/jobs_en_5k.csv")
```

```python
inputs = df_jobs[["Original title", "std_title_name"]].rename(columns={"Original title": "job title", "std_title_name": "standard title"}).to_dict("records")
```

```python
await check_title(text=str(inputs[0]))
```

```python
gpt_titles = []
for sample in tqdm(inputs):
    gpt_title = await check_title(str(sample))
    gpt_titles.append(gpt_title)
```

```python
df_jobs["gpt_title"] = gpt_titles
```

```python
df_jobs.to_csv("data/4k_jobs_std_titles.csv", index=False)
```

```python
(df_jobs["std_title_name"].str.lower() == df_jobs["gpt_title"].str.lower()).sum()/len(df_jobs)
```

```python
import json
import pandas as pd
from job_agents.generate_titles import generate_titles
```

```python
df = pd.read_csv("data/jobs_en_5k.csv")
```

```python
inputs = df[["Unique ID", "Original title"]].rename(columns={"Unique ID": "id", "Original title": "raw title"}).to_dict("records")
```

```python
def chunks(lst, n):
    for i in range(0, len(lst), n):
        yield lst[i:i + n]
```

```python
results, all_tokens = [], []
```

```python
for titles in chunks(inputs[3310:], 10):
    output, n_tokens = await generate_titles(str(titles))
    results += output
    all_tokens.append(n_tokens)
    print(f"processed {len(results)} titles")
```
